## 1. Variables and Arithmetic Expressions

A **variable** is a name that refers to a value. In Python, you create or update a variable using the assignment operator `=`.

Arithmetic expressions combine numbers and operators to calculate a result.

| Operator | Meaning | Example |
|---|---|---|
| `+` | Addition | `8 + 2` |
| `-` | Subtraction | `8 - 2` |
| `*` | Multiplication | `8 * 2` |
| `/` | Division | `8 / 2` |
| `//` | Floor division | `9 // 2` |
| `%` | Remainder (modulo) | `9 % 2` |
| `**` | Exponentiation | `3 ** 2` |

Python follows the usual order of operations: parentheses first, then exponentiation, multiplication/division, and finally addition/subtraction.

### Example

```python
price = 12
quantity = 3
total = price * quantity
print(total)
```

The expression calculates the total cost of three items.

## 4. Boolean Expressions and Logical Operators

A **Boolean expression** evaluates to either `True` or `False`.

Comparison operators are used to compare values:

| Operator | Meaning |
|---|---|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |

Logical operators combine or modify Boolean expressions:

- `and` is `True` only when both conditions are true.
- `or` is `True` when at least one condition is true.
- `not` reverses a Boolean value.

The modulo operator `%` returns the remainder of a division. It is useful for checking divisibility. For example, a number is even when its remainder after division by 2 is zero.

## 5. Conditional Statements (`if`, `elif`, `else`)

Conditional statements allow a program to choose what to do based on a condition.

The general structure is:

```python
if condition:
    # Instructions when the condition is true
elif another_condition:
    # Instructions when the second condition is true
else:
    # Instructions when none of the conditions is true
```

Python checks conditions from top to bottom. Once a condition is true, its corresponding block runs and the remaining branches are skipped.

Pay attention to indentation: the instructions belonging to a branch must be indented consistently.

## 6. Strings and String Methods

A **string** is a sequence of characters, usually written between single or double quotation marks.

Python provides several useful string operations and methods:

- `len(text)` returns the number of characters.
- `text[0]` accesses the first character.
- `text[-1]` accesses the last character.
- `text.upper()` returns an uppercase version.
- `text.lower()` returns a lowercase version.
- `text.strip()` removes whitespace from the beginning and end.
- `text.startswith(...)` checks how a string begins.
- `text.endswith(...)` checks how a string ends.

Strings are immutable: string methods return a new string rather than changing the original string in place.

## 8. The `range()` Function

The `range()` function generates a sequence of integers. It is commonly used with `for` loops.

It can receive one, two, or three arguments:

- `range(stop)` starts at 0 and stops before `stop`.
- `range(start, stop)` starts at `start` and stops before `stop`.
- `range(start, stop, step)` also specifies the increment between values.

The stop value is **not included**. The step can be positive or negative.

For example, `range(2, 8, 2)` represents the values 2, 4, and 6.

Use `range()` when you want a loop to process a regular sequence of integers:


This starts at 3, increases by 2, and stops before 10. A negative step can produce a descending sequence. Remember that the stop value is never included; this detail is important for the countdown, negative-sequence, and multiplication-table exercises.
